```python
from pathlib import Path

archivo = Path("config/settings.py")
lineas = archivo.read_text(encoding="utf-8").splitlines()

resultado = []
portal_encontrado = False

for linea in lineas:
    if linea.strip() == "'portal',":
        if portal_encontrado:
            continue
        portal_encontrado = True
    resultado.append(linea)

archivo.write_text("\n".join(resultado) + "\n", encoding="utf-8")

print("Duplicado eliminado.")
```

```python
from pathlib import Path

Path("portal/templates/portal").mkdir(parents=True, exist_ok=True)

Path("portal/views.py").write_text("""
from django.shortcuts import render

def inicio(request):
    return render(request, "portal/inicio.html")
""".strip() + "\n", encoding="utf-8")

Path("portal/urls.py").write_text("""
from django.urls import path
from . import views

urlpatterns = [
    path("", views.inicio, name="inicio"),
]
""".strip() + "\n", encoding="utf-8")

Path("portal/templates/portal/inicio.html").write_text("""
<!DOCTYPE html>
<html lang="es">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1">
    <title>Portal de Estudiantes</title>
    <style>
        body {
            font-family: Arial, sans-serif;
            background: #f2f5f9;
            margin: 0;
            color: #243247;
        }
        header {
            background: #17365d;
            color: white;
            padding: 25px;
            text-align: center;
        }
        main {
            max-width: 800px;
            margin: 45px auto;
            padding: 30px;
            background: white;
            border-radius: 12px;
            text-align: center;
        }
        a {
            display: inline-block;
            margin-top: 15px;
            padding: 12px 22px;
            background: #17365d;
            color: white;
            text-decoration: none;
            border-radius: 6px;
        }
    </style>
</head>
<body>
    <header>
        <h1>Portal de Estudiantes</h1>
        <p>Tu espacio de formación y documentos</p>
    </header>
    <main>
        <h2>Bienvenido</h2>
        <p>Desde aquí podrás consultar tus oportunidades y gestionar tus documentos.</p>
        <a href="/admin/">Acceso administrativo</a>
    </main>
</body>
</html>
""".strip() + "\n", encoding="utf-8")

print("Página de inicio creada correctamente.")
```
